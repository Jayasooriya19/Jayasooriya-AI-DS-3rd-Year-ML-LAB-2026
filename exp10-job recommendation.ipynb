```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_csv("Job Datsset.csv")
df.head()
```

```python
df.shape
df.columns
```

```python
df = df.fillna('')
job_requirements = df['Job_Requirements']
print(job_requirements.head())
```

```python
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(df['Job_Requirements'] )
print(tfidf_matrix.shape)
```

```python
user_input = input("Enter your skills: ")
user_vector = vectorizer.transform([user_input])
```

```python
similarity = cosine_similarity( user_vector, tfidf_matrix ).flatten()
top_jobs = similarity.argsort()[-5:][::-1]
```

```python
print("\n===== JOB RECOMMENDATIONS =====")
for i, index in enumerate(top_jobs):
    print("Recommendation", i + 1)
    print("Job ID :", df.iloc[index]['Job_ID'])
    print("Similarity Score :", round(similarity[index], 3))
    print()
```
